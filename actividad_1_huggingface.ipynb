## 2. Importar las clases y cargar el dataset

```python
import pandas as pd
from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer

loader = HFDatasetLoader(dataset_id="scikit-learn/iris", split="train")
df = loader.load()
print(f"Dataset cargado: {df.shape[0]} filas y {df.shape[1]} columnas")
display(df.head())
```

## 3. Transformación mínima

Se renombra `sepal_length` a `sepal_length_cm` cuando esa columna está presente. Renombrar una columna es una transformación real del DataFrame; no solo una visualización.

```python
if "sepal_length" in df.columns:
    df_transformado = df.rename(columns={"sepal_length": "sepal_length_cm"}).copy()
elif "sepal length (cm)" in df.columns:
    df_transformado = df.rename(columns={"sepal length (cm)": "sepal_length_cm"}).copy()
else:
    # Alternativa robusta: seleccionar las columnas disponibles y renombrar la primera.
    df_transformado = df.copy()
    if len(df_transformado.columns) > 0:
        primera_columna = df_transformado.columns[0]
        df_transformado = df_transformado.rename(columns={primera_columna: f"{primera_columna}_transformada"})

print("Columnas originales:", list(df.columns))
print("Columnas transformadas:", list(df_transformado.columns))
display(df_transformado.head())
```

## 4. Generar el archivo de resumen `.txt`

La clase recibe el DataFrame transformado. Por tanto, el archivo de resumen describe los datos después de la transformación.

```python
summarizer = DatasetSummarizer(df_transformado)
ruta_resumen = summarizer.save_summary("resumen_dataset.txt")
print(f"Resumen guardado en: {ruta_resumen}")
with open(ruta_resumen, "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## 5. Explicación del flujo

1. `HFDatasetLoader` carga el split `train` desde Hugging Face y lo convierte en un DataFrame de pandas.
2. El notebook renombra una columna para demostrar una transformación mínima.
3. `DatasetSummarizer` recibe el DataFrame transformado y genera `resumen_dataset.txt` con dimensiones, tipos, valores nulos, estadísticas numéricas y una muestra.
